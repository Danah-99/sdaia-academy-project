## 2. Import Libraries

```python
from ultralytics import YOLO
import cv2
```

## 3. Load a Pretrained YOLO Model

We start with a pretrained YOLO model so we can test the webcam before creating our own custom model.

```python
# Load a small pretrained YOLO model
model = YOLO("yolo11n.pt")
```

## 4. Open the Webcam

`0` normally refers to the computer's default camera.

```python
# Open the default webcam
cap = cv2.VideoCapture(0)

# Check that the camera opened successfully
if not cap.isOpened():
    print("Could not open the camera")
else:
    print("Camera opened successfully")
```

## 5. Real-Time YOLO Detection

The camera continuously provides frames. YOLO processes each frame and returns the detected objects.

**Press `Q` to stop the camera.**

```python
while True:
    # Read one frame from the camera
    ret, frame = cap.read() #frame the pic and ret = True → frame was captured successfully. False → something went wrong / no frame was captured

    # Stop if the frame could not be read
    if not ret:
        break

    # Run YOLO detection on the current frame
    results = model(frame, conf=0.25)

    # Draw bounding boxes and labels on the frame
    annotated_frame = results[0].plot() #first/default camera connected to the computer. .plot to draw detection

    # Display the detected frame
    cv2.imshow("YOLO Real-Time Detection", annotated_frame)

    # Press Q to stop
    if cv2.waitKey(1) & 0xFF == ord("q"): #1 millisecond.
        break

# Release the camera
cap.release()          # Stop using the camera
cv2.destroyAllWindows() # Close the OpenCV display window
```

```python
results = model("https://ultralytics.com/images/bus.jpg", conf=0.25)

results[0].show()
```

## 6. Understand the Pipeline

The complete process is:

**Webcam → Frame → YOLO → Detection → Display → Next Frame**

YOLO runs once for each captured frame.
